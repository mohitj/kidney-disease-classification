```python
import os
import random
import shutil

# Source folder containing 5000 images
source_folder = r"C:\Users\r105539\Downloads\archive (1)\CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone\CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone\Tumor"

# Destination folder where sampled images will be saved
destination_folder = r"C:\Users\r105539\Downloads\archive (1)\CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone\CT-KIDNEY-DATASET-Normal-Cyst-Tumor-Stone\Tumorpatient"

# Create destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

# Get all image files
image_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.gif', '.tiff')

all_images = [
    f for f in os.listdir(source_folder)
    if f.lower().endswith(image_extensions)
]

# Sort images (important for consistent grouping)
all_images.sort()

# Pick 1 random image from every 20 images
selected_images = []

for i in range(0, len(all_images), 10):
    batch = all_images[i:i+20]

    if batch:  # avoid empty batch
        random_image = random.choice(batch)
        selected_images.append(random_image)

# Copy selected images to destination folder
for image_name in selected_images:
    src_path = os.path.join(source_folder, image_name)
    dst_path = os.path.join(destination_folder, image_name)

    shutil.copy2(src_path, dst_path)

print(f"Total images found: {len(all_images)}")
print(f"Total sampled images saved: {len(selected_images)}")
print("Done!")
```
